```python
#4.Write a program that checks if a given word is a Python keyword.
import keyword

def is_python_keyword(word):
  """
  Checks if the given word is a reserved Python keyword.

  Args:
    word: The string to check.

  Returns:
    True if the word is a keyword, False otherwise.
  """
  return keyword.iskeyword(word)

# --- Example Usage ---
word1 = "if"
word2 = "my_variable"
word3 = "for"
word4 = "apple"

print(f"'{word1}' is a keyword: {is_python_keyword(word1)}")
print(f"'{word2}' is a keyword: {is_python_keyword(word2)}")
print(f"'{word3}' is a keyword: {is_python_keyword(word3)}")
print(f"'{word4}' is a keyword: {is_python_keyword(word4)}")

# You can also get a list of all current keywords:
print("\nList of all current Python keywords:")
print(keyword.kwlist)
```

```python
#5.Create a list and tuple in Python, and demonstrate how attempting to change an element works differently for each.
# A list (uses square brackets [])
my_list = [10, 20, 30, 40]

# A tuple (uses parentheses ())
my_tuple = (10, 20, 30, 40)

print(f"Original list: {my_list}")

# Change the second element (index 1) to 99
my_list[1] = 99

print(f"List after change: {my_list}")
# Output: List after change: [10, 99, 30, 40]

print(f"Original tuple: {my_tuple}")

# Attempt to change the second element (index 1) to 99
try:
    my_tuple[1] = 99
except TypeError as e:
    print(f"Error when trying to change tuple element: {e}")
# Output: Error when trying to change tuple element: 'tuple' object does not support item assignment
```

```python
#6.Write a function to demonstrate the behavior of mutable and immutable arguments.
def modify_list(my_list):
    """Appends an item to a list, demonstrating mutable behavior."""
    print(f"Inside function (before): {my_list} (id: {id(my_list)})")
    my_list.append(4)  # Modifies the list in place
    print(f"Inside function (after): {my_list} (id: {id(my_list)})")

# Demonstrate
original_list = [1, 2, 3]
print(f"Outside function (before): {original_list} (id: {id(original_list)})")
modify_list(original_list)
print(f"Outside function (after): {original_list} (id: {id(original_list)})")
def modify_integer(my_int):
    """Attempts to modify an integer, demonstrating immutable behavior."""
    print(f"Inside function (before): {my_int} (id: {id(my_int)})")
    my_int = my_int + 1  # Creates a new integer object
    print(f"Inside function (after): {my_int} (id: {id(my_int)})")

# Demonstrate
original_int = 10
print(f"Outside function (before): {original_int} (id: {id(original_int)})")
modify_integer(original_int)
print(f"Outside function (after): {original_int} (id: {id(original_int)})")
```

```python
#14.Write a program to reverse a string using a while loop.
def reverse_string_while_loop(input_string):
    """
    Reverses a string using a while loop.

    Args:
        input_string: The string to be reversed.

    Returns:
        The reversed string.
    """
    reversed_str = ""
    index = len(input_string) - 1  # Start index at the last character

    # Continue the loop as long as the index is 0 or greater
    while index >= 0:
        reversed_str += input_string[index]
        index -= 1  # Decrement the index to move leftwards through the string

    return reversed_str

# Example usage:
original_string = "Hello, World!"
reversed_result = reverse_string_while_loop(original_string)

print(f"Original string: {original_string}")
print(f"Reversed string: {reversed_result}")
```
